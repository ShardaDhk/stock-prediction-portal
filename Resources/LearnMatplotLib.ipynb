```python
import matplotlib.pyplot as plt   #pyplot is module of matplotlib libraray
import pandas as pd
import numpy as np
```

### Type of Plots

#### Line Plot

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 40]
```

```python
plt.plot(x, y);
plt.title("Simple Line Plot");
plt.xlabel('X-axis lable');
plt.ylabel("Y-axis lable");
plt.grid(True)
```

#### Bar chart

```python
month = ['jan', 'feb', 'Mar', 'apr']
revenue = [100, 200, 300, 500]
```

```python
plt.figure(figsize=(10, 6))
plt.bar(month, revenue)
plt.title("Monthly Revenue Bar Chart");
plt.xlabel('Months');
plt.ylabel('Revenue');
```

#### Histogram

```python
StudentsMarks = np.random.randint(0, 100, 50)
StudentsMarks
```

```python
plt.hist(StudentsMarks);
plt.xlabel('Marks')
plt.ylabel('No. of Students')
```

#### Scatter Plot

```python
heights = [65, 70, 68, 72, 66]
weights = [150, 160, 155, 180, 145]
```

```python
plt.scatter(heights, weights)
plt.title('Heights vs. Weights Scatter Plot')
plt.xlabel('Height (inches)')
plt.ylabel('weight (pounds)')
```

#### Legends

```python
days = [1, 2, 3, 4, 5]
original_price = [10, 20, 25, 30, 40]
predicted_price = [8, 21, 25, 28, 38]
```

```python
plt.plot(days, original_price, 'b-o', label='Original Price')
plt.plot(days, predicted_price,'r-o', label='Predicted Price')
plt.legend()
plt.title('plot with legends')
plt.xlabel('Weekdays')
plt.ylabel('Price')
```

#### Subplots

```python
# First Subplot - Original Price
plt.subplot(3, 1, 1)
plt.plot(days, original_price, 'b-o', label='Original Price')
plt.title('Original Price over time')
plt.xlabel('Weekdays')
plt.ylabel('Price')
plt.legend()

# Second Subplot - Predicted Price
plt.subplot(3, 1, 2)
plt.plot(days, predicted_price,'r-o', label='Predicted Price')
plt.title('Predicted Price over time')
plt.xlabel('Weekdays')
plt.ylabel('Price')
plt.legend()

# Third Subplot - Difference between Original & Predicted Price
price_difference = [o - p for o, p in zip(original_price, predicted_price)]
plt.subplot(3, 1, 3)
plt.bar(days, price_difference,color='purple', label='Price Difference')
plt.title('Difference between Original & Predicted Price')
plt.xlabel('Days')
plt.ylabel('Price Difference')
plt.legend()
plt.tight_layout()
```

#### Save the Figure

```python
plt.plot(days, original_price, 'b-o', label='Original Price')
plt.plot(days, predicted_price,'r-o', label='Predicted Price')
plt.legend()
plt.title('plot with legends')
plt.xlabel('Weekdays')
plt.ylabel('Price')
plt.savefig('example_prediction.png', dpi=300)  #save the image
```
